# EDA

Funcion con los metodos que  deseamos incluir  en el EDA inicial

```python
def resumen_eda(df):
    print("📊 Dimensiones del dataset")
    print(f"Filas: {df.shape[0]}")
    print(f"Columnas: {df.shape[1]}")
    print("-" * 40)

    print("🔎 Tipos de datos")
    print(df.dtypes)
    print("-" * 40)

    print("⚠️ Valores nulos por columna")
    nulos = df.isnull().sum()
    if (nulos > 0).any():
        print(nulos[nulos > 0].sort_values(ascending=False))
    else:
        print("No hay valores nulos en el DataFrame (0 en todas las columnas)")
    print("-" * 40)

    print("📌 Porcentaje de nulos por columna")
    porcentaje_nulos = df.isnull().mean() * 100
    if (porcentaje_nulos > 0).any():
        print(porcentaje_nulos[porcentaje_nulos > 0].sort_values(ascending=False))
    else:
        print("No hay valores nulos (0% en todas las columnas)")
    print("-" * 40)

    print("🔁 Registros duplicados")
    duplicados = df.duplicated().sum()
    print(f"Cantidad de duplicados: {duplicados}")
    print("-" * 40)

    print("📈 Estadísticas descriptivas")
    print(df.describe(include='number'))
    print("-" * 40)

    print("📈 Estadísticas descriptivas")
    try:
        print(df.describe(include='object'))
    except:
      print("No hay variables categóricas")
    print("-" * 40)
```

##Serie de Tiempo

```python
ven_mar_rrss.head()
```

```python
ven_mar_rrss["month"] = ven_mar_rrss["fecha_venta"].dt.month
```

```python
ven_mar_rrss["en_campanha"] = ( # en_campanha es true si se cumplen ambas condiciones
    (ven_mar_rrss["fecha_venta"] >= ven_mar_rrss["fecha_inicio"]) &
    (ven_mar_rrss["fecha_venta"] <= ven_mar_rrss["fecha_fin"])
)
```

```python
ven_mar_rrss_time = ven_mar_rrss.groupby(["month", "en_campanha"], as_index=False).agg(
    venta_neta_avg = ("venta_neta", "mean"),
    costo_sum = ("costo", "sum")
)
```

```python
ven_mar_rrss_time.head(3)
```

```python
# generamos la columna estado a partir de en_campanha
ven_mar_rrss_time["estado"] = ven_mar_rrss_time["en_campanha"].map({
    True: "En campaña",
    False: "Fuera de campaña"
})

# drop de la columna inicial
ven_mar_rrss_time.drop(columns=["en_campanha"], inplace=True)

# ordenamos
ven_mar_rrss_time.sort_values(by="month", ascending=True, inplace=True)
```

```python
ven_mar_rrss_time.head()
```

```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
ven_mar_rrss.head(2)
```

```python

kpi_ventas = (
    ven_mar_rrss
    .groupby("en_campanha", as_index=False)
    .agg(venta_neta_avg=("venta_neta", "mean"))
)

venta_prom_en_campanha = kpi_ventas.loc[
    kpi_ventas["en_campanha"] == True,
    "venta_neta_avg"
].iloc[0]

venta_prom_fuera_campanha = kpi_ventas.loc[
    kpi_ventas["en_campanha"] == False,
    "venta_neta_avg"
].iloc[0]


costo_marketing_rrss = df_mar_rrss["costo"].sum()

fig_kpis = make_subplots(
    rows=1,
    cols=3,
    specs=[[{"type": "indicator"}, {"type": "indicator"}, {"type": "indicator"}]]
)

fig_kpis.add_trace(
    go.Indicator(
        mode="number",
        value=venta_prom_en_campanha,
        title={"text": "Venta promedio<br>En campaña"},
        number={"prefix": "$", "valueformat": ",.2f"}
    ),
    row=1, col=1
)

fig_kpis.add_trace(
    go.Indicator(
        mode="number",
        value=venta_prom_fuera_campanha,
        title={"text": "Venta promedio<br>Fuera de campaña"},
        number={"prefix": "$", "valueformat": ",.2f"}
    ),
    row=1, col=2
)

fig_kpis.add_trace(
    go.Indicator(
        mode="number",
        value=costo_marketing_rrss,
        title={"text": "Costo marketing<br>RRSS"},
        number={"prefix": "$", "valueformat": ",.2f"}
    ),
    row=1, col=3
)

fig_kpis.update_layout(
    title="KPIs generales - Campañas RRSS",
    height=280
)

fig_kpis.show()
```

Conclusiones

A partir del análisis exploratorio realizado sobre los datasets ficticios de ventas, clientes y marketing, se pudieron aplicar distintas técnicas de preparación, transformación, análisis y visualización de datos utilizando Python.

1. Calidad y preparación de los datos

Se identificaron y trataron problemas de calidad en el dataset de ventas, incluyendo 35 registros duplicados y 2 registros con valores nulos en las columnas de precio y cantidad. También se transformaron los tipos de datos de precios y fechas para facilitar los cálculos y el análisis posterior. Los datasets de clientes y marketing no presentaron registros duplicados ni valores nulos en la revisión inicial.

2. Análisis del rendimiento comercial

Se calculó la venta neta por registro a partir del precio unitario y la cantidad vendida. Mediante agrupaciones y agregaciones, se analizaron los ingresos y las cantidades vendidas por producto, así como los ingresos por categoría. Este procedimiento permite identificar diferencias de rendimiento comercial y reconocer qué productos y categorías merecen un análisis más detallado.

3. Análisis de campañas de marketing

Se integraron los datos de ventas y campañas para explorar la relación entre los períodos de actividad de las campañas en redes sociales (RRSS) y el comportamiento de las ventas. Se utilizaron comparaciones estadísticas y visualizaciones para examinar las diferencias en los importes de venta y su distribución durante y fuera de los períodos de campaña.

Sin embargo, los resultados de esta comparación deben considerarse preliminares, ya que es necesario validar la relación entre las ventas y las campañas para evitar contabilizar varias veces una misma venta cuando existen múltiples campañas asociadas a un producto.

4. Visualización y comunicación de resultados

Se desarrollaron visualizaciones estáticas e interactivas utilizando Matplotlib, Seaborn y Plotly. Los gráficos permitieron explorar distribuciones, relaciones entre variables, evolución temporal y diferencias entre productos, facilitando la interpretación de los datos desde una perspectiva comercial.

5. Limitaciones y oportunidades de mejora

Al tratarse de datos ficticios, los resultados no representan necesariamente el comportamiento de un negocio real. Además, para evaluar correctamente la efectividad de las campañas, sería necesario validar la asociación entre cada venta y las campañas correspondientes, revisar el cálculo de los indicadores de retorno y considerar otros factores que puedan influir en las ventas.

Como mejora futura, se propone profundizar el análisis del comportamiento de los clientes, incorporar indicadores comerciales adicionales y desarrollar un dashboard que facilite el seguimiento de las principales métricas.
